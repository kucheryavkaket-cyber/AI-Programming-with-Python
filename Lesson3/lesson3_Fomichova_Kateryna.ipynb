### Висновки за результатами описової статистики

Для стовпця Age середній вік пацієнтів становить 74.91 року, медіана  75 років. Близькість цих значень свідчить про те, що центральні показники розподілу майже збігаються. Вік пацієнтів перебуває в діапазоні від 60 до 90 років, стандартне відхилення становить 8.99. Виявлено дві моди: 68 і 88 років.

Для стовпця BMI середнє значення становить 27.66, медіана  27.82. Мінімальне значення  15.01, максимальне  39.99, розмах  24.98, стандартне відхилення  7.22. Виразна мода відсутня, оскільки всі значення унікальні.

```python
print(round(df["Age"].corr(df["BMI"]), 2))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 5))

r = df["Age"].corr(df["BMI"])

ax.scatter(df["Age"], df["BMI"], s=5, alpha=0.25, color="#0072B2")

ax.set_title(f"Кореляція Age та BMI: r = {r:.2f}")
ax.set_xlabel("Age (вік)")
ax.set_ylabel("BMI (індекс маси тіла)")

print("Age і BMI:", round(r, 2))

plt.tight_layout()
plt.show()
```

### Аналіз кореляції

Для перевірки взаємозв'язку між віком пацієнтів Age та індексом маси тіла BMI розраховано коефіцієнт кореляції. Отримане значення становить -0.02, що свідчить про практично відсутній лінійний зв'язок між цими показниками.

Кореляція не доводить причинно-наслідковий зв'язок, оскільки на показники можуть впливати інші фактори, а наявність статистичного зв'язку сама по собі не означає, що один показник спричиняє зміни іншого.

Перевірка викидів методом IQR для стовпця BMI

```python
Q1 = df["BMI"].quantile(0.25)
Q3 = df["BMI"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Нижня межа:", lower)
print("Верхня межа:", upper)
print("Кількість можливих викидів:", ((df["BMI"] < lower) | (df["BMI"] > upper)).sum())
```
